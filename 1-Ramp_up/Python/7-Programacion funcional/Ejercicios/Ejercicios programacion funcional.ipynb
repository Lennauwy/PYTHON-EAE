# Ejercicios Programación funcional
## Ejercicio 1
Utiliza `reduce` para concatenar estas palabras en una frase
```Python
ejer_1 = ("to", "be", "or", "not", "to", "be", "that", "is", "the", "question")
```

````python
from functools import reduce

ejer_1 = ("to", "be", "or", "not", "to", "be", "that", "is", "the", "question")

# Using reduce to join the words with a space in between
frase = reduce(lambda x, y: x + " " + y, ejer_1)

print(frase)
# Output: to be or not to be that is the question
````

## Ejercicio 2
Obtén una nueva tupla con el cuadrado de cada elemento de la siguiente tupla

```Python
ejer_2 = (1,2,3,4,5)
```

````python
ejer_2 = (1, 2, 3, 4, 5)

# Use map to calculate the square and convert the result back to a tuple
resultado = tuple(map(lambda x: x**2, ejer_2))

print(resultado)
# Output: (1, 4, 9, 16, 25)
````

## Ejercicio 3
Filtra aquellos elementos que sean multiplos de 3 o de 5
```Python
ejer_3 = (1,2,3,4,5,15,21,22,33,34,35)
```

````python
ejer_3 = (1, 2, 3, 4, 5, 15, 21, 22, 33, 34, 35)

# Use filter with a lambda function for the conditions
# We check if (n % 3 == 0) OR (n % 5 == 0)
result = tuple(filter(lambda n: n % 3 == 0 or n % 5 == 0, ejer_3))

print(result)
# Output: (3, 5, 15, 21, 33, 35)
````

## Ejercicio 4
Convierte la siguiente tupla de pesetas a euros
```Python
ejer_4 = (1000, 8000, 20000000, 40000000)
```

````python
ejer_4 = (1000, 8000, 20000000, 40000000)

# Apply the conversion rate (pesetas / 166.386) to each element
euros = tuple(map(lambda p: round(p / 166.386, 2), ejer_4))

print(euros)
# Output: (6.01, 48.08, 120202.42, 240404.84)
````

## Ejercicio 5
Convierte la siguiente lista de desconocidos a mayusculas
```Python
ejer_5 = ("Arya", "John", "Robb", "Bran", "Sansa", "Rickon")
```

````python
ejer_5 = ("Arya", "John", "Robb", "Bran", "Sansa", "Rickon")

# Use map to apply the .upper() method to each string
names_upper = tuple(map(lambda name: name.upper(), ejer_5))

print(names_upper)
# Output: ('ARYA', 'JOHN', 'ROBB', 'BRAN', 'SANSA', 'RICKON')
````

## Ejercicio 6
Realiza un conversor de dias de la semana en texto a numerico (1-7), usando `map`.
```Python
ejer_6 = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")
```

````python
ejer_6 = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo")

# Create a mapping dictionary
day_to_num = {
    "Lunes": 1, "Martes": 2, "Miércoles": 3, "Jueves": 4, 
    "Viernes": 5, "Sábado": 6, "Domingo": 7
}

# Use map to convert the text to numbers
# We use .get() to avoid errors if a day is misspelled
result = tuple(map(lambda day: day_to_num.get(day), ejer_6))

print(result)
# Output: (1, 2, 3, 4, 5, 6, 7)
````

## Ejercicio 7
Utiliza `map` para multiplicar elemento a elemento estas dos tuplas
```Python
ejer_7_1 = (2, 5, 8)
ejer_7_2 = (6, 3, 2)
```

````python
ejer_7_1 = (2, 5, 8)
ejer_7_2 = (6, 3, 2)

# Use map with two arguments in the lambda function
result = tuple(map(lambda x, y: x * y, ejer_7_1, ejer_7_2))

print(result)
# Output: (12, 15, 16)
````

## Ejercicio 8
Crea una tupla nueva con todos los coches que no sean VW :)

```Python
ejer_8 = ("VW", "Audi", "Renault", "VW", "BMW")
```

````python
ejer_8 = ("VW", "Audi", "Renault", "VW", "BMW")

# Use filter to keep items that are NOT "VW"
# != is the operator for "not equal to"
result = tuple(filter(lambda car: car != "VW", ejer_8))

print(result)
# Output: ('Audi', 'Renault', 'BMW')
````

## Ejercicio 9
Dadas las siguientes fechas en formato string, quedate con aquellas del año 2020

```Python
ejer_9 = ("2019-04-08", "2020-10-10", "2020-01-22", "2019-07-13", "2019-02-01")
```

````python
ejer_9 = ("2019-04-08", "2020-10-10", "2020-01-22", "2019-07-13", "2019-02-01")

# Use filter to keep strings that begin with "2020"
result = tuple(filter(lambda date: date.startswith("2020"), ejer_9))

print(result)
# Output: ('2020-10-10', '2020-01-22')
````

## Ejercicio 10
Filtra aquellos strings que sean [palíndromos](https://es.wikipedia.org/wiki/Pal%C3%ADndromo)
```Python
ejer_10 = ("php", "w3r", "Python", "abcd", "Java", "aaa")
```

````python
ejer_10 = ("php", "w3r", "Python", "abcd", "Java", "aaa")

# Use filter to keep strings that are equal to their reverse
result = tuple(filter(lambda word: word == word[::-1], ejer_10))

print(result)
# Output: ('php', 'aaa')
````
