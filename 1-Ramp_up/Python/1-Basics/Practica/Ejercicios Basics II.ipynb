## Exercise 2
Write a program to convert degrees to radians. You must use `input`. Remember that the conversion is done through

radians = degrees*(pi/180)

````python
import math

# 1. Get user input and convert it to a float
degrees = float(input("Enter degrees: "))

# 2. Perform the conversion using the formula
# math.pi provides the value of pi (3.14159...)
radians = degrees * (math.pi / 180)

# 3. Print the result
print(radians)
````

## Exercise 6
Same program as in 5, but this time there must be three inputs and two outputs. One output should tell us if all are equal, and the other if at least two inputs are equal

````python
# 1. Collect three user inputs
val1 = input("Enter first value: ")
val2 = input("Enter second value: ")
val3 = input("Enter third value: ")

# 2. Check if ALL are equal (A must equal B AND B must equal C)
all_equal = (val1 == val2 == val3)

# 3. Check if AT LEAST TWO are equal (A=B OR B=C OR A=C)
any_two_equal = (val1 == val2) or (val2 == val3) or (val1 == val3)

# 4. Output the results
print("Are all three equal?", all_equal)
print("Are at least two equal?", any_two_equal)
````

## Exercise 7
Create a program that takes two inputs. It must check if their sum is equal, greater than, or less than 10

````python
# 1. Collect two numeric inputs
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# 2. Calculate the sum
total = num1 + num2

# 3. Check the conditions
if total == 10:
    print("The sum is equal to 10")
elif total > 10:
    print("The sum is greater than 10")
else:
    print("The sum is less than 10")
````

## Exercise 10
For the following string you are asked to print the following cases on screen

    "Early birds catch the worm"

1. Convert it all to uppercase
2. Convert it all to lowercase
3. Convert to uppercase only the initial of the words
4. Create a list by dividing it by its spaces
5. Replace commas `,` with semicolons `;`
6. Remove lowercase `a` letters

````python
text = "Early birds catch the worm"

# 1. Convert it all to uppercase
print(text.upper())

# 2. Convert it all to lowercase
print(text.lower())

# 3. Convert to uppercase only the initial of the words (Title Case)
print(text.title())

# 4. Create a list by dividing it by its spaces
print(text.split(" "))

# 5. Replace commas with semicolons
# Note: This string has no commas, but the method remains the same
print(text.replace(",", ";"))

# 6. Remove lowercase 'a' letters
# We "remove" them by replacing 'a' with an empty string ""
print(text.replace("a", ""))
````

## Exercise 11
1. Create a list with 3 numeric elements
2. Add a fourth element
3. Calculate the sum of all
4. Delete the second element of the list
5. Add another element at position 3 of the list
6. Create another list with 4 elements and concatenate it to the one you already had.

````python
# 1. Create a list with 3 numeric elements
my_list = [10, 20, 30]

# 2. Add a fourth element (adds to the end)
my_list.append(40)

# 3. Calculate the sum of all
total_sum = sum(my_list)
print("Sum of list:", total_sum)

# 4. Delete the second element of the list
# Note: Python uses 0-based indexing, so the second element is at index 1
del my_list[1]

# 5. Add another element at position 3 of the list
# The .insert(index, value) method adds it at a specific spot
my_list.insert(2, 99)

# 6. Create another list with 4 elements and concatenate it
second_list = [1, 2, 3, 4]
final_list = my_list + second_list

# Let's see the final result
print("Final list:", final_list)
````
