## Exercise 5
Create a Celsius to Fahrenheit converter. The converter must receive the user input and return the conversion.

The conversion of degrees follows the following formula: F = C × (9 / 5) + 32

```python
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = celsius * (9 / 5) + 32
print("The temperature in Fahrenheit is:", fahrenheit)
```
