## Exercise 1
Write a function that converts numbers from 1 to 7 into names of the days of the week. The function will consist of a single numeric argument and a string type output

````python
def get_day_name(day_number):
    """
    Converts a number (1-7) to the corresponding day of the week.
    """
    days = {
        1: "Monday",
        2: "Tuesday",
        3: "Wednesday",
        4: "Thursday",
        5: "Friday",
        6: "Saturday",
        7: "Sunday"
    }
    
    # Use .get() to return a message if the number is out of range
    return days.get(day_number, "Invalid day number. Please use 1-7.")

# Example usage:
print(get_day_name(1))  # Output: Monday
print(get_day_name(5))  # Output: Friday
````

## Exercise 2
In exercise 8 of loops, we created an inverted pyramid, whose number of floors was determined by user input. Create a function that replicates the behavior of the pyramid, and uses a single input parameter of the function to determine the number of rows of the pyramid, that is, removes the input statement.

````python
def print_inverted_pyramid(rows):
    """
    Prints a countdown pattern starting from 'rows' down to 1.
    """
    # Outer loop determines the starting number of each line
    for i in range(rows, 0, -1):
        # Inner loop counts down from i to 1
        for j in range(i, 0, -1):
            print(j, end=" ")
        # Move to the next line after finishing the row
        print()

# Example usage:
print_inverted_pyramid(5)
````

## Exercise 3
Write a function that compares two numbers. The function has two arguments and there are three possible outputs: they are equal, the first is greater than the second, or the second is greater than the first

````python
def compare_numbers(a, b):
    """
    Compares two numbers and returns their relationship as a string.
    """
    if a == b:
        return "They are equal"
    elif a > b:
        return "The first is greater than the second"
    else:
        return "The second is greater than the first"

# Example usage:
print(compare_numbers(10, 5))   # First is greater
print(compare_numbers(3, 8))    # Second is greater
print(compare_numbers(7, 7))    # They are equal
````

## Exercise 4
Write a function that is a letter counter. In the first argument you have to introduce a text, and the second that is the letter to count. The function must return an integer with the number of times that letter appears, both uppercase and lowercase

````python
def count_letter(text, letter):
    """
    Counts the occurrences of a specific letter in a text, 
    ignoring case sensitivity.
    """
    # Convert everything to lowercase to ensure a fair count
    text_lower = text.lower()
    letter_lower = letter.lower()
    
    # Use the built-in .count() method for strings
    total = text_lower.count(letter_lower)
    
    return total

# Example usage:
my_text = "Python is powerful and Popular"
target = "p"
result = count_letter(my_text, target)

print(f"The letter '{target}' appears {result} times.")
````

## Exercise 5
Write a function that has a single argument, a string. The output of the function must be a dictionary with the count of all the letters of that string.

````python
def frequency_counter(text):
    """
    Returns a dictionary with the count of each character in the string.
    """
    counts = {}
    
    for char in text:
        # Check if the character is already in our dictionary
        if char in counts:
            counts[char] += 1
        else:
            # If it's the first time seeing the character, initialize it at 1
            counts[char] = 1
            
    return counts

# Example usage:
result = frequency_counter("hello world")
print(result)
````

## Exercise 6
Write a function that adds or removes elements in a list. The function needs the following arguments:
* list: the list where elements will be added or removed
* command: "add" or "remove"
* element: By default it is None.

````python
def modify_list(target_list, command, element=None):
    """
    Adds or removes an element from a list based on the command.
    """
    if command == "add":
        if element is not None:
            target_list.append(element)
        else:
            print("Error: You must provide an element to add.")
            
    elif command == "remove":
        if element in target_list:
            target_list.remove(element)
        elif element is None:
            print("Error: You must provide an element to remove.")
        else:
            print(f"Error: Element '{element}' not found in the list.")
            
    else:
        print("Invalid command. Please use 'add' or 'remove'.")
    
    return target_list

# Example usage:
my_list = [1, 2, 3]

# Adding an element
modify_list(my_list, "add", 4)
print("After adding:", my_list)

# Removing an element
modify_list(my_list, "remove", 2)
print("After removing:", my_list)
````

## Exercise 7
Create a function that receives an arbitrary number of words, and returns a complete sentence, separating the words with spaces.

````python
def make_sentence(*words):
    """
    Combines an arbitrary number of words into a single string.
    """
    # Use .join() to combine the tuple of words with spaces
    sentence = " ".join(words)
    
    return sentence

# Example usage:
print(make_sentence("This", "is", "a", "Python", "exercise"))
# Output: This is a Python exercise

print(make_sentence("Hello", "everyone"))
# Output: Hello everyone
````

## Exercise 8
Write a program that obtains the nth number of the [Fibonacci sequence](https://en.wikipedia.org/wiki/Fibonacci_number). You must create a recursive function with a single argument.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibonacci(n):
    """
    Returns the nth number of the Fibonacci sequence using recursion.
    """
    # Base cases: if n is 0 or 1, return n
    if n <= 1:
        return n
    else:
        # Recursive step: sum of the two previous numbers
        return fibonacci(n - 1) + fibonacci(n - 2)

# Testing the examples provided
print(f"fibonacci(2) = {fibonacci(2)}") # Output: 1
print(f"fibonacci(4) = {fibonacci(4)}") # Output: 3
print(f"fibonacci(5) = {fibonacci(5)}") # Output: 5
print(f"fibonacci(8) = {fibonacci(8)}") # Output: 21
````

## Exercise 9
Define in a single cell the following functions:
* Function that calculates the area of a square
* Function that calculates the area of a triangle
* Function that calculates the area of a circle

In another cell, calculate the area of:
* Two circles with radius 10 + a triangle with base 3 and height 7
* A square with side = 10 + 3 circles (one with radius = 4 and the other two with radius = 6) + 5 triangles with base = 2 + height = 4

````python
import math

def area_square(side):
    return side ** 2

def area_triangle(base, height):
    return (base * height) / 2

def area_circle(radius):
    return math.pi * (radius ** 2)
````

````python
# Calculation 1: Two circles (r=10) + a triangle (b=3, h=7)
calc_1 = (2 * area_circle(10)) + area_triangle(3, 7)

# Calculation 2: Square (s=10) + 3 circles (r=4, r=6, r=6) + 5 triangles (b=2, h=4)
calc_2 = area_square(10) + \
         area_circle(4) + (2 * area_circle(6)) + \
         (5 * area_triangle(2, 4))

print(f"Result 1: {calc_1:.2f}")
print(f"Result 2: {calc_2:.2f}")
````
