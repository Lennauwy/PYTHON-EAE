## Exercise 1
We are a company that has several appliance stores. We need a program to manage the stores, see the sales they have had, how many employees there are, etc... For this we are going to model the store in a class, which will have the following characteristics:
* Class name: "Store"
* Common attributes:
    * Type: "Appliances"
    * Open: True
* Attributes specific to each store:
    * Name - String
    * Address - String
    * Number of employees - int
    * Sales last 3 months - List of 3 numbers
* Method to calculate sales from all months, returning a number with all sales.
* Method that calculates the average sales of the last months, per employee
* Method that returns in a string the name of the store, along with its address.
* Method that gets the sales from the last month.
* Method to give a sales projection in case X money has been invested in marketing for that store. Where X is the input parameter. If X is less than 1000, sales for the last 3 months would have been (1.2 x sales), if greater than or equal to 1000, sales would have been (1.5 x sales). The method rewrites the sales attribute for the last 3 months, and also returns the sales with the increase.

It is requested:
1. Implement the "Store" class
2. Create three stores with invented data
3. Check in at least one of them, everything implemented in the store class (its attributes, average sales, sales/employee...)
3. Calculate the sales from the last month of all stores. For this use the `for` loop
4. Print on screen the names of the stores whose address contains the string "Avenue"
5. I'm sure you've already done this section... Document the class :)

```python
class Store:
    """
    A class to represent an appliance store.

    Attributes:
        type (str): The general category of the store (Default: "Appliances").
        open (bool): The operational status (Default: True).
        name (str): Name of the specific branch.
        address (str): Physical location.
        employees (int): Number of staff members.
        sales_history (list): Sales figures for the last 3 months.
    """
    # Common attributes (Class attributes)
    store_type = "Appliances"
    is_open = True

    def __init__(self, name, address, employees, sales_history):
        # Instance attributes
        self.name = name
        self.address = address
        self.employees = employees
        self.sales_history = sales_history

    def total_sales(self):
        """Returns the sum of all sales in the history."""
        return sum(self.sales_history)

    def sales_per_employee(self):
        """Calculates average sales per employee over the tracked period."""
        if self.employees == 0: return 0
        return self.total_sales() / self.employees

    def get_info(self):
        """Returns the name and address formatted as a string."""
        return f"Store: {self.name} | Location: {self.address}"

    def last_month_sales(self):
        """Returns the most recent month's sales (last item in list)."""
        return self.sales_history[-1]

    def marketing_projection(self, investment):
        """
        Adjusts sales history based on marketing investment.
        Investment < 1000 -> 20% increase.
        Investment >= 1000 -> 50% increase.
        """
        multiplier = 1.5 if investment >= 1000 else 1.2
        
        # Rewriting the sales attribute using a list comprehension
        self.sales_history = [sale * multiplier for sale in self.sales_history]
        return self.sales_history
```

```python
# Creating instances
store_a = Store("ElectroCity", "123 Main Avenue", 5, [10000, 12000, 15000])
store_b = Store("TechHub", "456 Tech Street", 10, [20000, 25000, 30000])
store_c = Store("ValueMart", "789 Park Avenue", 3, [5000, 6000, 5500])

# Checking Store A
print(store_a.get_info())
print(f"Total Sales: {store_a.total_sales()}")
print(f"Sales per Employee: {store_a.sales_per_employee():.2f}")
```

```python
all_stores = [store_a, store_b, store_c]
total_last_month = 0

for store in all_stores:
    total_last_month += store.last_month_sales()

print(f"Combined sales for last month: {total_last_month}")
```

```python
print("Stores located on an Avenue:")
for store in all_stores:
    if "Avenue" in store.address:
        print(f"- {store.name}")
```

## Exercise 2
In this exercise we are going to implement a *Dog* class in Python. The class has the following characteristics:
* Things we know for sure a dog has
    * It has 4 legs
    * 2 ears
    * 2 eyes
    * A speed of 0. By default, the dog is stopped
* When initialized:
    * The dog will be of a certain breed
    * By default it will have "Brown" fur, unless otherwise stated.
    * By default it will have no owner, unless otherwise stated.
    
* It will also have a method called walk, which has an input argument (speed_increase). This value will be added to the speed the dog already had.
* It needs another method (stop), where we will set the speed to 0.
* Another method called "bark", which will have an input argument, and the output will be the following string: "WOOF!" + the input argument.


It is requested:
* Implement the *Dog* class
* Create an object of type *Dog*, without an owner
* Check that all its attributes are correct
* Test that it walks, and check its speed
* Stop it
* Document the *Dog* class

```python
class Dog:
    """
    A class to represent a Dog.
    
    Attributes:
        legs (int): Number of legs (Default: 4).
        ears (int): Number of ears (Default: 2).
        eyes (int): Number of eyes (Default: 2).
        speed (int): Current movement speed (Default: 0).
        breed (str): The dog's breed.
        fur (str): Color of the fur (Default: 'Brown').
        owner (str): Name of the owner (Default: None).
    """
    # Shared characteristics (Class attributes)
    legs = 4
    ears = 2
    eyes = 2

    def __init__(self, breed, fur="Brown", owner=None):
        # Specific characteristics (Instance attributes)
        self.breed = breed
        self.fur = fur
        self.owner = owner
        self.speed = 0

    def walk(self, speed_increase):
        """Increases the current speed of the dog."""
        self.speed += speed_increase
        print(f"Walking... Speed increased by {speed_increase}. Current speed: {self.speed}")

    def stop(self):
        """Resets the dog's speed to 0."""
        self.speed = 0
        print("Stopped. Speed is now 0.")

    def bark(self, message):
        """Returns a string combining a bark and a message."""
        return f"WOOF! {message}"
```

```python
# Create an object of type Dog without an owner
my_dog = Dog(breed="Golden Retriever")

# Check attributes
print("--- Attribute Check ---")
print(f"Breed: {my_dog.breed}")   # Golden Retriever
print(f"Fur: {my_dog.fur}")       # Brown (Default)
print(f"Owner: {my_dog.owner}")   # None (Default)
print(f"Legs: {my_dog.legs}")     # 4
print(f"Initial Speed: {my_dog.speed}") # 0

# Test walking
print("\n--- Movement Test ---")
my_dog.walk(5)  # Speed becomes 5
my_dog.walk(3)  # Speed becomes 8

# Test stopping
my_dog.stop()   # Speed becomes 0

# Test barking
print("\n--- Interaction Test ---")
print(my_dog.bark("I'm hungry!"))
```
