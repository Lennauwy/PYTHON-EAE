```python
# Setup from Exercise 1
# Index: 0:age, 1:name, 2:height, 3:Windows?, 4:None, 5:series, 6:clothes, 7:info_dict
lista_examen = [25, "Lennard Nack", 1.83, True, None, 
                ["Breaking Bad", "The Bear", "Succession", "Dark", "Severance"], 
                ("Nike", "Levi's", "Adidas"), 
                {"ciudad": "Madrid", "portal": 12}]

# 1. SCREEN DISPLAY (No loops/custom functions)
print(f"Name: {lista_examen[1]}, Age: {lista_examen[0]}")
print(f"2nd series: {lista_examen[5][1]}, Penultimate: {lista_examen[5][-2]}")
print(f"Footwear brand: {lista_examen[6][2]}")

if lista_examen[2] < 1.80:
    print(f"Result (Height < 1.80): {lista_examen[6][2]}")
else:
    print(f"Result (Height >= 1.80): {lista_examen[6][0]}")

vowels = "AEIOU"
if lista_examen[1][0].upper() not in vowels and lista_examen[7]["ciudad"][0].upper() not in vowels:
    print(f"{lista_examen[7]['ciudad']}, {lista_examen[1]}")
else:
    print(lista_examen[7]["ciudad"])

# 2. FOR LOOPS
print("\n--- All elements ---")
for item in lista_examen:
    print(item)

print("\n--- Reversed ---")
for item in lista_examen[::-1]:
    print(item)

print("\n--- Odd indices of nested collections ---")
for item in lista_examen:
    # Check if item is a list or tuple (ordered collections)
    if type(item) == list or type(item) == tuple:
        for i in range(len(item)):
            if i % 2 != 0: # Check for odd index
                print(item[i])

print("\n--- City Check ---")
lista_de_ciudades = ["Barcelona", "Madrid"]
found = False
for city in lista_de_ciudades:
    if city == lista_examen[7]["ciudad"]:
        print(city)
        found = True
if not found:
    print("city not found")

# 3. WHILE LOOP
series_borradas = []
while len(lista_examen[5]) > 2:
    series_borradas.append(lista_examen[5].pop()) # Removes from the tail

# 4. ADD TO THE MIDDLE
# The list now has 2 items left. Index 1 is the middle.
middle = len(lista_examen[5]) // 2
for s in series_borradas:
    lista_examen[5].insert(middle, s)

print("\nFinal series list:", lista_examen[5])
```

### Ejercicio 4 (2 puntos)

1. Crea una clase "Humano" que tenga 4 atributos: "ojos"(int default 2), "edad"(int), "altura"(float) y "nombre"(str). Además, debe contar con dos métodos: 

- "cumplir_edad": al llamar a este método se debe sumar 1 al atributo edad del humano y que retorne la edad actualizada. 
- "registro_nombre" que recibe parametro "nuevo_nombre": al llamar a esta función con un parámetro de entrada, este método debe cambiar el nombre del humano por el "nuevo_nombre" y sin retornar nada

2. Crea un humano llamado "Cojoncio" que tiene el doble de tu de edad (a partir de la variable `lista_examen`) y mide 10 centímetros menos que tu altura (a partir de la variable `lista_examen`). A "Cojoncio" nunca le gustó su nombre. Por eso, usa el método correspondiente para cambiarle su nombre por "Goku". Además, hoy es su cumpleaños por lo que usa el método correspondiente para aumentar su edad. Muestra tanto el nombre como la edad antes y después ejecutar los métodos.


3. Documenta la clase con el estilo de Google


```python
# Setup: Your list from previous exercises
lista_examen = [25, "Lennard Nack", 1.83, True, None, 
                ["Breaking Bad", "The Bear", "Succession", "Dark", "Severance"], 
                ("Nike", "Levi's", "Adidas"), 
                {"ciudad": "Madrid", "portal": 12}]

# 1. Create the Humano class
class Humano:
    # Constructor mapping the attributes. 'ojos' is set to a default of 2.
    def __init__(self, edad, altura, nombre, ojos=2):
        self.ojos = ojos
        self.edad = edad
        self.altura = altura
        self.nombre = nombre
        
    # Method to increase age and return it
    def cumplir_edad(self):
        self.edad += 1
        return self.edad
    
    # Method to change the name (returns nothing)
    def registro_nombre(self, nuevo_nombre):
        self.nombre = nuevo_nombre

# 2. Instantiate the object and execute methods

# Calculate Cojoncio's starting stats based on your lista_examen
edad_cojoncio = lista_examen[0] * 2          # Double your age (25 * 2 = 50)
altura_cojoncio = lista_examen[2] - 0.10     # 10 cm less than your height in meters (1.83 - 0.10)

# Create the object
cojoncio = Humano(edad=edad_cojoncio, altura=altura_cojoncio, nombre="Cojoncio")

# Display state BEFORE executing methods
print("--- BEFORE ---")
print(f"Name: {cojoncio.nombre}")
print(f"Age: {cojoncio.edad}")

# Call the methods to modify the object's state
cojoncio.registro_nombre("Goku")
cojoncio.cumplir_edad()

# Display state AFTER executing methods
print("\n--- AFTER ---")
print(f"Name: {cojoncio.nombre}")
print(f"Age: {cojoncio.edad}")
```
