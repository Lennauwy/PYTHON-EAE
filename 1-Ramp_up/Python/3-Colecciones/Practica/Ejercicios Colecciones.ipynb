## Exercise 5
Given the following list, delete all values equal to 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: Do not try to delete elements from the list you are iterating. Make a copy with `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]

# Make a copy to iterate safely
for value in ejer_5.copy():
    if value == 3:
        ejer_5.remove(3)

print(ejer_5)
````

1. Create a tuple with 3 elements
2. Create another tuple with one element and check its type
3. Create a tuple with elements of different types
4. Print to screen the first and last element of the tuple from section 3. Use `len` for the last one
5. Add an element to the tuple from section 3.
6. Delete an element from the tuple in section 5, which is roughly in the middle.
7. Convert the tuple from section 5 into a list

````python
# 1. Create a tuple with 3 elements
tup1 = (10, 20, 30)

# 2. Create another tuple with one element and check its type
# NOTE: A comma is required for a single-element tuple!
tup2 = (5,)
print("Type of tup2:", type(tup2))

# 3. Create a tuple with elements of different types
tup3 = ("Apple", 42, 3.14, True)

# 4. Print first and last element (using len)
print("First:", tup3[0])
print("Last:", tup3[len(tup3) - 1])

# 5. Add an element to the tuple
# Since tuples are immutable, we must concatenate two tuples
tup3 = tup3 + ("New Element",)

# 6. Delete an element from the middle
# We convert to a list, remove, then convert back
temp_list = list(tup3)
temp_list.pop(len(temp_list) // 2) # Removes the middle element
tup3 = tuple(temp_list)

# 7. Convert the tuple from section 5 into a list
final_list = list(tup3)
print("Final list:", final_list)
````

## Exercise 9

3. How many elements does the `ejer_9` tuple have?
> ```ejer_9 = (3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3, 5, 2, 4, 7, 9, 4, 2, 4, 3, 3, 4, 6, 7)```

## Exercise 12
Convert the following list into a dictionary
> ```ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: You can use two control variables in a `for` loop. One for each value of the tuples.

````python
ejer_12 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
my_dict = {}

# Using two control variables: 'key' for the letter, 'value' for the number
for key, value in ejer_12:
    my_dict[key] = value

print(my_dict)
# Output: {'x': 3, 'y': 2, 'z': 1}
````

## Exercise 13
1. Create an ascending sorted list with the dictionary keys
2. Create another list sorted in descending order with the values
3. Add a new key/value
4. Find the key with value 2 in the dictionary
5. Iterate the key and value of the dictionary with a single for loop
> ```ejer_13 = {4:78, 2:98, 8:234, 5:29}```

````python
ejer_13 = {4: 78, 2: 98, 8: 234, 5: 29}

# 1. Create an ascending sorted list with the dictionary keys
sorted_keys = sorted(ejer_13.keys())
print("Sorted keys (asc):", sorted_keys)

# 2. Create another list sorted in descending order with the values
sorted_values_desc = sorted(ejer_13.values(), reverse=True)
print("Sorted values (desc):", sorted_values_desc)

# 3. Add a new key/value
ejer_13[10] = 500

# 4. Find the key with value 2 (Wait! Value 2 doesn't exist, but let's look for a value)
# If you meant the key for value 98:
target_value = 98
found_key = None
for key, value in ejer_13.items():
    if value == target_value:
        found_key = key
        break
print(f"Key with value {target_value}:", found_key)

# 5. Iterate the key and value of the dictionary with a single for loop
print("Iterating dictionary:")
for key, value in ejer_13.items():
    print(f"Key: {key}, Value: {value}")
````

## Exercise 14
Merge both dictionaries. To do this, use `update`
> ```ejer_14_1 = {1: 11, 2: 22}```
>
> ```ejer_14_2 = {3: 33, 4: 44}```

````python
ejer_14_1 = {1: 11, 2: 22}
ejer_14_2 = {3: 33, 4: 44}

# Merge ejer_14_2 into ejer_14_1
ejer_14_1.update(ejer_14_2)

print(ejer_14_1)
# Output: {1: 11, 2: 22, 3: 33, 4: 44}
````

## Exercise 16
Multiply all values of the dictionary
> ```ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}```

````python
ejer_16 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}

# Initialize the result variable to 1
total_product = 1

# Iterate through the dictionary values and multiply
for value in ejer_16.values():
    total_product *= value

print("The product of all values is:", total_product)
# Output: 19326120
````

## Exercise 17
1. Create a set with three elements
2. Add a fourth
3. Delete the last element added
4. Delete element 10, if it is present. Use `discard()`

````python
# 1. Create a set with three elements
my_set = {10, 20, 30}

# 2. Add a fourth
my_set.add(40)

# 3. Delete the last element added
# Note: Since sets are unordered, we can't target "the last one" by index. 
# We target the value specifically.
my_set.remove(40)

# 4. Delete element 10 if present using discard()
my_set.discard(10)

print(my_set)
````
